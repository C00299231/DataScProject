```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("kanchana1990/nasa-exoplanet-archive-intelligence")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

filename = r"nasa_exoplanet_intelligence.csv"

df = pd.read_csv(
    path + "\\" + filename,
    header=0
)
```

```python
print("dataset shape:", df.shape)
display(df.columns.tolist())
```

```python
# does radius match mass?
x = df["planet_radius_earth"]
y = df["planet_mass_earth"]

p = plt.figure()
#p.set_figwidth(27)
plt.scatter(x, y, s=6, alpha=0.2)
#print(df)
```

```python
plt.hist(df["planet_radius_earth"].dropna(), bins=50)
plt.xlabel("Planet Radius (Earth radii)")
plt.ylabel("Number of Planets")
plt.show()
```

```python
df[df["planet_radius_earth"].between(12, 14)]["planet_type"].value_counts()
```

```python
for planet_type, group in df.groupby("planet_type"):
    plt.scatter(
        group["planet_radius_earth"],
        group["planet_mass_earth"],
        label=planet_type,
        s=10,
        alpha=0.5
    )

plt.xlabel("Planet Radius (Earth radii)")
plt.ylabel("Planet Mass (Earth masses)")
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 6))

for planet_type, group in df.groupby("planet_type"):
    ax.scatter(
        group["planet_radius_earth"],
        group["planet_mass_earth"],
        label=planet_type,
        s=12,
        alpha=0.5,
    )

ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("earth radius")
ax.set_ylabel("earth mass")
ax.legend(title="Planet type", markerscale=2)
plt.tight_layout()
plt.show()
```
