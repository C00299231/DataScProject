```python
from astroquery.ipac.nexsci.nasa_exoplanet_archive import NasaExoplanetArchive
import pandas as pd

table = NasaExoplanetArchive.query_criteria_async(table="pscomppars", select="*")
df = table.to_table().to_pandas()
```

```python
print(df.shape)
```

```python
display(df.columns.tolist())
```

```python
cols = [
    # names
    "pl_name",
    "hostname",

    # system (no of stars, planets)
    "sy_snum",
    "sy_pnum",
    "sy_dist", # distance from earth

    # discovery
    "discoverymethod",
    "disc_year",
    "disc_facility",

    # planet parameters
    "pl_controv_flag",
    "pl_eqt", # equilibrium temp
    "pl_insol", # insolation flux

    "pl_orpper", # orbit period
    "pl_orbeccen", # orbit eccentricity
    "pl_orbsmax", # semimajor axis

    "pl_rade", # earth radii
    "pl_bmasse", # earth masses
    "pl_bmassprov", # mass provenance

    # stellar parameters
    "st_teff", # effective temperature
    "st_rad", # radius
    "st_mass", # mass
    "st_met", # metallicity
    "st_logg", # surface gravity
    "st_age", # age

    # positional
    #"ra",
    #"dec",
]

cols = [c for c in cols if c in df.columns]
slim = df[cols].copy()
slim.to_csv("exoplanets_slim_2026-10-15.csv", index=False)
```
