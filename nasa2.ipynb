```python
import pandas as pd
import matplotlib.pyplot as plt

filename = r"exoplanets_slim_2026-10-15.csv"

df = pd.read_csv(
    filename,
    header=0
)

df = df.rename(columns={
    "pl_name": "pl_name",
    "hostname": "host_name",

    "sy_snum": "sys_num_star",
    "sy_pnum": "sys_num_planet",
    "sy_dist": "sys_dist_e",

    "discoverymethod": "disc_method",
    #"disc_year": "",
    #"disc_facility": "",

    #"pl_controv_flag": "",
    "pl_eqt": "pl_temp",
    "pl_insol": "pl_insol_flux",

    "pl_orpper": "pl_orb_period",
    "pl_orbeccen": "pl_orb_eccen",
    "pl_orbsmax": "pl_orb_semimajor_axis",

    "pl_rade": "pl_radius_e",
    "pl_bmasse": "pl_mass_e",
    "pl_bmassprov": "pl_mass_prov",

    "st_teff": "st_temp",
    "st_rad": "st_radius",
    #"st_mass": "",
    "st_met": "st_metal",
    "st_logg": "st_gravity",
    #"st_age": "",
    })

print(df.shape)
display(df.dtypes)
```

```python
fig, ax = plt.subplots(figsize=(9, 6))

noSinMass = (df["pl_mass_prov"] != "Msini") & (df["pl_mass_prov"] != "Msin(i)/sin(i)")
onlyMass = df["pl_mass_prov"] == "Mass"
onlySiniSiniMass = df["pl_mass_prov"] == "Msin(i)/sin(i)"

for mass_provenance, group in df[onlyMass].groupby("pl_mass_prov"):
    ax.scatter(
        group["pl_radius_e"],
        group["pl_mass_e"],
        label=mass_provenance,
        s=10,
        alpha=0.1,
    )

ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("earth radius")
ax.set_ylabel("earth mass")
plt.tight_layout()
plt.legend()
plt.show()
```

```python
x = df["pl_insol_flux"]
y = df["pl_temp"]

p = plt.figure()
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Planet Insolation Flux")
plt.ylabel("Planet Equilibrium Temperature")
plt.scatter(x, y, alpha=0.1, s=5)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
model = LinearRegression(fit_intercept=True)
df.dropna(inplace=True)
x = df[["pl_insol_flux"]]
y = df["pl_temp"]
model.fit(x, y)
xfit = np.linspace(0, 100000, 1000)
yfit = model.predict(xfit[:, np.newaxis])
plt.xscale("log")
plt.yscale("log")
plt.scatter(x, y, alpha=0.01, s=8)
plt.plot(xfit, yfit)
print(model.score(x, y))
```
